```python
import pandas as pd

data = {
    "jampengamatan": ["07.00", "08.00", "09.00", "10.00", "11.00", "12.00", "13.00", "14.00", "15.00", "16.00"],
    "volumekendaraan": [120, 185, 150, 110, 135, 160, 145, 125, 175, 190],
}

df = pd.DataFrame(data)
print(df)
```

```python
total = df['volumekendaraan'].sum()
print("total volume kendaraan:", total)
```

```python
rata_rta = df['volumekendaraan'].mean()
print("rata rata volume kendaraan:", rata_rta)
```

```python
tertinggi = df['volumekendaraan'].max()
terendah = df['volumekendaraan'].min()
print("volume kendaraan tertinggi:", tertinggi)
print("volume kendaraan terendah:", terendah)
```

```python
di_atas_rata = df[df['volumekendaraan'] > rata_rta]
print("volume kendaraan di atas rata-rata:")
print(di_atas_rata)
```
